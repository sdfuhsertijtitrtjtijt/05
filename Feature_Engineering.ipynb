Feature engineering is the process of transforming raw data into meaningful features that can improve the quality of machine learning models.
Health diagnosis combined with feature engineering plays an important role in healthcare data analysis. By preprocessing patient information, creating meaningful features, selecting relevant attributes, and applying machine learning techniques, useful predictive models can be developed to support medical research and clinical decision-making. Such models are supportive tools and cannot replace professional medical diagnosis.











```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

from sklearn.preprocessing import (
    StandardScaler,
    MinMaxScaler,
    LabelEncoder
)

from sklearn.feature_selection import (
    mutual_info_classif
)

import warnings
warnings.filterwarnings('ignore')

sns.set_style("whitegrid")

print("All libraries imported successfully!")
```

```python

np.random.seed(42)

n = 500

df = pd.DataFrame({
    'Patient_ID': range(1, n+1),
    'Age': np.random.randint(18, 85, n),
    'Gender': np.random.choice(
        ['Male', 'Female'], n
    ),
    'BMI': np.random.normal(27, 5, n).round(2),
    'Glucose': np.random.normal(115, 30, n).round(2),
    'Blood_Pressure': np.random.normal(125, 18, n).round(2),
    'Cholesterol': np.random.normal(205, 35, n).round(2),
    'Heart_Rate': np.random.randint(55, 120, n),
    'Smoking': np.random.choice(
        ['Yes', 'No'], n
    ),
    'Physical_Activity': np.random.choice(
        ['Low', 'Medium', 'High'], n
    ),
    'Family_History': np.random.choice(
        ['Yes', 'No'], n
    )
})

# Generate an illustrative binary outcome
risk = (
    (df['Age'] > 50).astype(int) +
    (df['BMI'] > 30).astype(int) +
    (df['Glucose'] > 140).astype(int) +
    (df['Blood_Pressure'] > 140).astype(int) +
    (df['Cholesterol'] > 240).astype(int) +
    (df['Smoking'] == 'Yes').astype(int)
)

df['Diagnosis'] = np.where(
    risk >= 3, 'Higher Risk', 'Lower Risk'
)

print("Healthcare Dataset Created Successfully")
display(df.head(10))

print("Dataset Shape:", df.shape)
```

```python

print("Dataset Information")
df.info()

print("\nStatistical Summary")
display(df.describe())

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Records")
print(df.duplicated().sum())

print("\nDiagnosis Distribution")
print(df['Diagnosis'].value_counts())

print("\nData Types")
print(df.dtypes)
```

```python

# Check missing values
print("Before Cleaning:")
print(df.isnull().sum().sum())

# Fill missing numerical values
num_cols = df.select_dtypes(
    include=np.number
).columns

df[num_cols] = df[num_cols].fillna(
    df[num_cols].median()
)

# Fill categorical missing values
cat_cols = df.select_dtypes(
    include='object'
).columns

for col in cat_cols:
    df[col] = df[col].fillna(
        df[col].mode()[0]
    )

# Remove duplicates
df = df.drop_duplicates()

print("After Cleaning:")
print("Missing Values:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())

```

```python

# 1. Age group
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0, 30, 45, 60, 100],
    labels=['Young', 'Adult', 'Middle Age', 'Senior']
)

# 2. BMI category
df['BMI_Category'] = pd.cut(
    df['BMI'],
    bins=[0, 18.5, 25, 30, 100],
    labels=[
        'Underweight',
        'Normal',
        'Overweight',
        'Obese'
    ]
)

# 3. Glucose category
df['Glucose_Category'] = pd.cut(
    df['Glucose'],
    bins=[0, 100, 126, 300],
    labels=['Normal Range', 'Elevated', 'High']
)

# 4. Blood pressure category
df['BP_Category'] = pd.cut(
    df['Blood_Pressure'],
    bins=[0, 120, 130, 140, 250],
    labels=[
        'Lower',
        'Elevated',
        'Stage 1 Range',
        'Stage 2 Range'
    ]
)

# 5. Cholesterol category
df['Cholesterol_Category'] = pd.cut(
    df['Cholesterol'],
    bins=[0, 200, 240, 400],
    labels=['Desirable', 'Borderline', 'High']
)

# 6. Lifestyle indicator
df['Smoking_Indicator'] = (
    df['Smoking'] == 'Yes'
).astype(int)

# 7. Family history indicator
df['Family_History_Indicator'] = (
    df['Family_History'] == 'Yes'
).astype(int)

# 8. Interaction feature
df['BMI_Glucose_Interaction'] = (
    df['BMI'] * df['Glucose']
)

print("Feature Engineering Completed!")
display(df.head(10))
```

```python

# Label encoding
le = LabelEncoder()

df['Gender_Encoded'] = le.fit_transform(df['Gender'])

df['Diagnosis_Encoded'] = le.fit_transform(
    df['Diagnosis']
)

# One-hot encoding
df = pd.get_dummies(
    df,
    columns=['Physical_Activity'],
    prefix='Activity',
    dtype=int
)

# Standard scaling
scaler = StandardScaler()

df['BMI_Standardized'] = scaler.fit_transform(
    df[['BMI']]
)

# Min-Max scaling
minmax = MinMaxScaler()

df['Glucose_Normalized'] = minmax.fit_transform(
    df[['Glucose']]
)

print("Encoding and Scaling Completed!")
display(df.head())
```

```python

cross = pd.crosstab(
    df['BMI_Category'],
    df['Diagnosis']
)

cross.plot(
    kind='bar',
    stacked=True,
    figsize=(9,5),
    colormap='viridis'
)

plt.title("BMI Category vs Diagnosis")
plt.xlabel("BMI Category")
plt.ylabel("Number of Patients")
plt.show()
```

```python

age_avg = df.groupby(
    'Age_Group',
    observed=False
)['Glucose'].mean()

age_avg.plot(
    kind='area',
    figsize=(9,5),
    color='lightgreen',
    alpha=0.7
)

plt.title("Average Glucose Across Age Groups")
plt.xlabel("Age Group")
plt.ylabel("Average Glucose")
plt.show()
```

```python

features = [
    'Age',
    'BMI',
    'Glucose',
    'Blood_Pressure',
    'Cholesterol',
    'Heart_Rate',
    'Smoking_Indicator',
    'Family_History_Indicator'
]

X = df[features]
y = df['Diagnosis_Encoded']

mi_scores = mutual_info_classif(
    X,
    y,
    random_state=42
)

importance = pd.Series(
    mi_scores,
    index=features
).sort_values()

importance.plot(
    kind='barh',
    figsize=(9,6),
    color='coral'
)

plt.title("Feature Importance Using Mutual Information")
plt.xlabel("Mutual Information Score")
plt.show()
```

```python

fig = px.sunburst(
    df,
    path=['Diagnosis', 'Gender', 'BMI_Category'],
    title='Healthcare Diagnosis Hierarchy'
)

fig.show()
```

```python

radar_data = df.groupby(
    'Diagnosis'
)[['Age', 'BMI', 'Glucose',
   'Blood_Pressure', 'Cholesterol']].mean()

# Normalize the group averages for visualization
radar_norm = (
    radar_data - radar_data.min()
) / (
    radar_data.max() - radar_data.min() + 1e-9
)

categories = radar_norm.columns.tolist()
angles = np.linspace(
    0, 2*np.pi, len(categories), endpoint=False
).tolist()

angles += angles[:1]

plt.figure(figsize=(8,8))
ax = plt.subplot(111, polar=True)

for index, row in radar_norm.iterrows():
    values = row.tolist()
    values += values[:1]

    ax.plot(
        angles,
        values,
        linewidth=2,
        label=index
    )

    ax.fill(angles, values, alpha=0.1)

ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories)
plt.title("Healthcare Feature Radar Chart")
plt.legend(loc='upper right', bbox_to_anchor=(1.3,1.1))
plt.show()
```
